```python
from ai_sdk import AISdk, AISdkConfig

import os

config = AISdkConfig.from_env()  # reads AI_SDK_HOST and AI_SDK_TOKEN
client = AISdk.from_config(config)
```

```python
from IPython.display import Markdown, display

conversation_id = None
buffer = ""
handle = display(Markdown(""), display_id=True)

for event in client.agent().stream("How do I compute the CLV?"):
    if not conversation_id:
        conversation_id = event.conversation_id
        print("Conversation ID:", conversation_id)
    if event.type == "content":
        buffer += event.content + "\n\n"
        handle.update(Markdown(buffer))
```
